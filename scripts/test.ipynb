```python
from PIL import Image, ExifTags, ImageCms
```

```python
img = Image.open('/home/scott/Downloads/originals_South Africa 2025__DSC0237-Edit copy 2.jpg')
```

```python
exif = img.getexif()
```

```python

for key, val in exif.items():
    tag_name = ExifTags.TAGS[key]
    if tag_name == "ExifOffset":
        print(f"key: '{key}', tag name: '{tag_name}', value: '{val}'")
        sub_ifd = img.getexif().get_ifd(key)
        for sub_key, sub_val in sub_ifd.items():
            sub_tag = ExifTags.TAGS.get(sub_key)
            print(f"       sub_key: '{sub_key}', sub tag name: '{sub_tag}', value: '{sub_val}'")
    else:
        print(f"key: '{key}', tag name: '{tag_name}', value: '{val}'")
```
